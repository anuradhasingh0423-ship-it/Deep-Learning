**Deep Learning**

it is a subset of Machine Learning that uses neural networks with multiple hidden layers to automatically learn patterns and features from large amounts of data.

It is inspired by the human brain, where neurons process information in layers to make decisions.

Deep Learning is a machine learning technique that uses multi-layered neural networks to learn complex patterns and make predictions from data.

## Networks

1. ANN  >> artificial

2. CNN >> convolutional nn

3. RNN >> recurrent neural network



### ANN

>> consist of 3 layers

>> all layers contain unit

>> each unit cpntains summation and activation fn  , summation = Wx+c  >> W is weight, x is input , c is constant




### Activation Function

>> linear , it is used in output layer for regression

>> sigmoid, it is used in output layer for binary classification (Yes / No)

>> softmax, it is used in output layer for multiclass classification (Blood grp)

>> ReLu , it is used in hidden layer

### Forward Propagation


it is a process where input is passed to hidden layer by adding some weights
to the output layers

### Backward Propagation

when the weights are updated for better output by repeating the forward propagation is known as backward propagation

### Epochs

An epoch is one complete pass of the entire training dataset through the neural network.
it is complete cycle of forward and backward propagation

### Dropouts

Dropout is a regularization technique where a fraction of neurons are randomly turned off (dropped) during training to prevent overfitting.

it is used for prevention overfitting

in dropouts we drop certain amount of units

### **Early stopping**

if the performance is dropping continuously than early stopping will stop the epochs.

Early stopping is a technique that automatically stops training when the model's performance on validation data stops improving, preventing overfitting.

### **Optimizers**

An optimizer is an algorithm that updates the model’s weights to minimize the loss function during training.

it is used to find the optimum of learning rate

### **Learning Rate**

The learning rate is a hyperparameter that controls how much the model’s weights change during each update while minimizing the loss.

### Types of Optimizers

1. Adam

2. **Gradient descent** >> Gradient Descent is an optimization algorithm used to minimize the loss function by iteratively updating model parameters in the direction of the negative gradient.

3. **AdaGrade**  >> AdaGrad is an adaptive optimization algorithm that automatically adjusts the learning rate for each parameter, giving larger updates to infrequent features and smaller updates to frequent ones.
